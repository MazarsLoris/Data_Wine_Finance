**Vérification des types et formats**

```python
data.dtypes
```

```python
STRING_COLUMNS = [
    "Id Commande", "Id Client", "Id Magasin Commande", "Canal Vente", 
    "Statut Commande", "Id Employé", "Nom Complet Employe", 
    "Poste Employé", "Nom Magasin", "Ville Magasin", "Type Magasin", 
    "Id Detail", "Id Produit", "Nom Produit", "Id Facture", 
    "Statut Paiement Facture"
]

DATE_COLUMNS = ["Date Commande", "Date Comptable"]

NUMERIC_COLUMNS = ["Quantite", "Prix Unitaire", "Sous-Total Produit", "Prix Achat Produit"]

def format_data(df):
    # Transformation en texte
    for col in STRING_COLUMNS:
        if col in df.columns:
            df[col] = df[col].astype(str)

    # Transformation en Dates
    for col in DATE_COLUMNS:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors="coerce")

    # Transformation en Numériques
    for col in NUMERIC_COLUMNS:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")
            
    return df

data = format_data(data)

# Vérification rapide des types
data.info()
```
